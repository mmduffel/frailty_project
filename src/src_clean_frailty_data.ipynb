```python
#Upload the raw dataset into Google Colab
from google.colab import files

#Import pandas for data cleaning and manipulation
import pandas as pd

uploaded = files.upload()

#Load the raw dataset into a DataFrame
df = pd.read_csv("raw_frailty_data.csv", encoding="latin1")

#Preview the first five rows
df.head()
```

```python
#Look over dataset structure and types
df.info()
```

```python
# Display summary statistics for the numerical variables
df.describe()
```

```python
# Display the data type of each variable
df.dtypes
```

```python
# View column names in the dataset
df.columns
```

```python
# Remove extra spaces from column names
df.columns = df.columns.str.strip()

# Verify that extra spaces were removed from column names
print(df.columns)
```

```python
# Remove whitespace from numeric column values
df["Height"] = df["Height"].astype(str).str.replace(r"\s+", "", regex=True)
df["Weight"] = df["Weight"].astype(str).str.replace(r"\s+", "", regex=True)
df["Age"] = df["Age"].astype(str).str.replace(r"\s+", "", regex=True)
df["Grip strength"] = df["Grip strength"].astype(str).str.replace(r"\s+", "", regex=True)
```

```python
# Convert numeric columns from object to numeric data types
df["Height"] = pd.to_numeric(df["Height"])
df["Weight"] = pd.to_numeric(df["Weight"])
df["Age"] = pd.to_numeric(df["Age"])
df["Grip strength"] = pd.to_numeric(df["Grip strength"])
```

```python
# Verify that the data types were successfully converted
df.dtypes
```

```python
# Clean Frailty column before encoding
df["Frailty"] = df["Frailty"].astype(str).str.strip().str.upper()
```

```python
# Verify Frailty values are standardized to Y and N
df["Frailty"].unique()
```

```python
# Check for missing values in each column
df.isnull().sum()
```

```python
# Check for duplicate rows
df.duplicated().sum()
```

```python
# Verify the dataset shape before and after removing missing values
print("Before:", df.shape)
df = df.dropna()
print("After:", df.shape)
```

```python
# Save cleaned dataset as a CSV file
df.to_csv("frailty_data_clean.csv", index=False)
```
